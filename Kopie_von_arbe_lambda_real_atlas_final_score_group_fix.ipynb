# ARBE Lambda Star Structural Decision Notebook

## What is different here

This version is designed for production handoff:

- real atlas field names are used directly
- request loading is configurable through a **single mapping cell**
- requests can come from **CSV** or **JSON**
- requests may contain either:
  - direct LAB targets, or
  - atlas reference ids that should be resolved to LAB through the atlas
- all final outputs remain **atlas-only**

```python
from pathlib import Path
import json
import math
import itertools
import pandas as pd
import numpy as np

# -------- paths --------
# Filled by the embedded bootstrap cell above.
ATLAS_JSON_PATH = Path(ATLAS_JSON_PATH)
SPECTRA_BIN_PATH = Path(SPECTRA_BIN_PATH)
REQUEST_PATH = Path(REQUEST_PATH)

# -------- mode switches --------
DECISION_MODE = True
STRICT_ATLAS_ONLY = True
FAIL_ON_NON_ATLAS_IN_DECISION = True

# -------- search / ranking controls --------
TOP_K_PER_TARGET = 8
TOP_GROUPS_TO_EXPORT = 50
MAX_TARGETS_FOR_GROUP_ENUMERATION = 6

# Prefer structural logic as primary decision policy.
USE_LEXICOGRAPHIC_DECISION = True

# -------- request schema adapter --------
# Edit aliases here to match your real request file columns.
REQUEST_ALIASES = {
    "target_id": ["target_id", "slot", "name", "target", "label"],
    "lab_L": ["target_l", "lab_L", "L", "L*", "lstar"],
    "lab_a": ["target_a", "lab_a", "a", "a*", "astar"],
    "lab_b": ["target_b", "lab_b", "b", "b*", "bstar"],
    "requested_reference_id": ["requested_reference_id", "reference_id", "atlas_id", "reference", "id"],
}
```

## Atlas loader

The atlas file is stored in **columnar JSON** form. The `fields` object contains one array per field.

```python
def load_columnar_atlas_json(path: Path) -> tuple[pd.DataFrame, dict]:
    with open(path, "r", encoding="utf-8") as f:
        raw = json.load(f)

    if raw.get("schema") != "hlc-atlas-lambda-star-v4-columnar":
        raise ValueError(f"Unexpected schema: {raw.get('schema')}")

    fields = raw["fields"]
    df = pd.DataFrame(fields)

    meta = {
        "schema": raw.get("schema"),
        "sort_order": raw.get("sort_order"),
        "stats": raw.get("stats", {}),
        "wavelengths_nm_10nm": raw.get("wavelengths_nm_10nm", []),
        "spectra_format": raw.get("spectra_format", {}),
        "version": raw.get("version"),
        "plugin_target": raw.get("plugin_target"),
    }
    return df, meta

atlas_df, atlas_meta = load_columnar_atlas_json(ATLAS_JSON_PATH)
print(atlas_meta["schema"])
print(f"rows={len(atlas_df):,}, cols={len(atlas_df.columns)}")
display(pd.Series(atlas_df.columns, name="atlas_columns"))
```

```python
REQUIRED_ATLAS_COLUMNS = [
    "id",
    "lab_L",
    "lab_a",
    "lab_b",
    "lambda_v2_nm",
    "lambda_ee_nm",
    "delta_lambda_nm",
    "mu2_nm2",
    "sigma_star_nm",
    "mu3_nm3",
]

missing = [c for c in REQUIRED_ATLAS_COLUMNS if c not in atlas_df.columns]
if missing:
    raise ValueError(f"Atlas is missing required columns: {missing}")

SCENARIO_DRIFT_COLUMNS = [c for c in atlas_df.columns if c.startswith("de00_from_d50__")]
REFRESH_COLUMNS = [c for c in atlas_df.columns if c.startswith("refresh__")]

print("Scenario drift columns:", SCENARIO_DRIFT_COLUMNS)
print("Refresh columns:", REFRESH_COLUMNS)
```

## Optional spectra file validation

This checks whether the external spectra binary matches the shape declared in the atlas metadata.

```python
def validate_spectra_file(bin_path: Path, meta: dict):
    spectra_meta = meta.get("spectra_format", {})
    expected_shape = tuple(spectra_meta.get("shape", []))
    dtype = np.dtype(spectra_meta.get("dtype", "float32"))
    if not bin_path.exists():
        return {"exists": False}
    arr = np.memmap(bin_path, dtype=dtype, mode="r")
    expected_size = int(np.prod(expected_shape))
    return {
        "exists": True,
        "declared_shape": expected_shape,
        "actual_value_count": int(arr.size),
        "matches_declared_shape": int(arr.size) == expected_size,
    }

spectra_check = validate_spectra_file(SPECTRA_BIN_PATH, atlas_meta)
spectra_check
```

## Request loader and schema adapter

This layer converts your request file into a standard internal schema:

- `target_id`
- `lab_L`
- `lab_a`
- `lab_b`
- optional `requested_reference_id`

If LAB values are missing but `requested_reference_id` is present, LAB will be looked up from the atlas.

```python
def load_request_file(path: Path) -> pd.DataFrame:
    suffix = path.suffix.lower()
    if suffix == ".csv":
        return pd.read_csv(path)
    if suffix == ".json":
        with open(path, "r", encoding="utf-8") as f:
            raw = json.load(f)
        if isinstance(raw, list):
            return pd.DataFrame(raw)
        if isinstance(raw, dict) and "fields" in raw:
            return pd.DataFrame(raw["fields"])
        if isinstance(raw, dict):
            return pd.DataFrame(raw)
    raise ValueError(f"Unsupported request file type: {suffix}")

def resolve_alias(df: pd.DataFrame, aliases: list[str]) -> str | None:
    cols_lower = {c.lower(): c for c in df.columns}
    for alias in aliases:
        if alias in df.columns:
            return alias
        if alias.lower() in cols_lower:
            return cols_lower[alias.lower()]
    return None

def standardize_request_df(df: pd.DataFrame, alias_map: dict, atlas_df: pd.DataFrame) -> pd.DataFrame:
    out = pd.DataFrame()

    for target_name, aliases in alias_map.items():
        found = resolve_alias(df, aliases)
        if found is not None:
            out[target_name] = df[found]

    # Backward compatibility for older request adapters.
    if "reference_id" in out.columns and "requested_reference_id" not in out.columns:
        out = out.rename(columns={"reference_id": "requested_reference_id"})

    # If the only identifier-like column is a generic `id`, prefer it as atlas reference id.
    if "requested_reference_id" not in out.columns and "id" in df.columns:
        has_explicit_target_id = "target_id" in df.columns
        has_any_lab = any(c in df.columns for c in ["lab_L", "lab_a", "lab_b", "L", "a", "b", "L*", "a*", "b*"])
        if not has_explicit_target_id or not has_any_lab:
            out["requested_reference_id"] = df["id"]

    if "target_id" not in out.columns:
        out["target_id"] = [f"T{i+1}" for i in range(len(df))]

    # Try atlas lookup if the request is id-based.
    if "requested_reference_id" in out.columns:
        out["requested_reference_id"] = out["requested_reference_id"].astype(str).str.strip()

    lab_cols_present = all(c in out.columns for c in ["lab_L", "lab_a", "lab_b"])
    if not lab_cols_present and "requested_reference_id" in out.columns:
        atlas_lookup = atlas_df[["id", "lab_L", "lab_a", "lab_b"]].rename(columns={"id": "requested_reference_id"})
        out = out.merge(atlas_lookup, on="requested_reference_id", how="left", validate="many_to_one")

    required_after_adapt = ["target_id", "lab_L", "lab_a", "lab_b"]
    missing = [c for c in required_after_adapt if c not in out.columns]
    if missing:
        raise ValueError(
            "Request file could not be standardized. Missing required columns after alias resolution: "
            f"{missing}.\nCurrent columns: {list(df.columns)}"
        )

    out["target_id"] = out["target_id"].astype(str).str.strip()
    for c in ["lab_L", "lab_a", "lab_b"]:
        out[c] = pd.to_numeric(out[c], errors="coerce")

    if out[["lab_L", "lab_a", "lab_b"]].isna().any().any():
        bad = out[out[["lab_L", "lab_a", "lab_b"]].isna().any(axis=1)]
        raise ValueError(f"Some requests do not resolve to valid LAB values:\n{bad}")

    return out

requests_raw_df = load_request_file(REQUEST_PATH)
targets_df = standardize_request_df(requests_raw_df, REQUEST_ALIASES, atlas_df)
display(targets_df)
```

```python
if len(targets_df) > MAX_TARGETS_FOR_GROUP_ENUMERATION:
    raise ValueError(
        f"This notebook is configured for at most {MAX_TARGETS_FOR_GROUP_ENUMERATION} targets per run. "
        f"Received {len(targets_df)}."
    )

target_ids = targets_df["target_id"].tolist()
print("Targets:", target_ids)
```

## Perceptual selection metrics

This section ranks atlas references by perceptual fit under D50.

```python
def ciede2000(L1, a1, b1, L2, a2, b2):
    # Sharma et al. 2005 implementation
    L1, a1, b1, L2, a2, b2 = map(float, (L1, a1, b1, L2, a2, b2))
    avg_L = (L1 + L2) / 2.0
    C1 = math.sqrt(a1*a1 + b1*b1)
    C2 = math.sqrt(a2*a2 + b2*b2)
    avg_C = (C1 + C2) / 2.0
    G = 0.5 * (1 - math.sqrt((avg_C**7) / (avg_C**7 + 25**7))) if avg_C != 0 else 0.0
    a1p = (1 + G) * a1
    a2p = (1 + G) * a2
    C1p = math.sqrt(a1p*a1p + b1*b1)
    C2p = math.sqrt(a2p*a2p + b2*b2)
    avg_Cp = (C1p + C2p) / 2.0

    def hp_fun(x, y):
        if x == 0 and y == 0:
            return 0.0
        h = math.degrees(math.atan2(y, x))
        return h + 360 if h < 0 else h

    h1p = hp_fun(a1p, b1)
    h2p = hp_fun(a2p, b2)

    dLp = L2 - L1
    dCp = C2p - C1p

    if C1p * C2p == 0:
        dhp = 0.0
    else:
        if abs(h2p - h1p) <= 180:
            dhp = h2p - h1p
        elif h2p <= h1p:
            dhp = h2p - h1p + 360
        else:
            dhp = h2p - h1p - 360

    dHp = 2 * math.sqrt(C1p * C2p) * math.sin(math.radians(dhp / 2.0))

    avg_Lp = (L1 + L2) / 2.0
    avg_Cp = (C1p + C2p) / 2.0

    if C1p * C2p == 0:
        avg_hp = h1p + h2p
    else:
        if abs(h1p - h2p) <= 180:
            avg_hp = (h1p + h2p) / 2.0
        elif (h1p + h2p) < 360:
            avg_hp = (h1p + h2p + 360) / 2.0
        else:
            avg_hp = (h1p + h2p - 360) / 2.0

    T = (
        1
        - 0.17 * math.cos(math.radians(avg_hp - 30))
        + 0.24 * math.cos(math.radians(2 * avg_hp))
        + 0.32 * math.cos(math.radians(3 * avg_hp + 6))
        - 0.20 * math.cos(math.radians(4 * avg_hp - 63))
    )
    d_ro = 30 * math.exp(-(((avg_hp - 275) / 25) ** 2))
    Rc = 2 * math.sqrt((avg_Cp**7) / (avg_Cp**7 + 25**7)) if avg_Cp != 0 else 0.0
    Sl = 1 + ((0.015 * ((avg_Lp - 50) ** 2)) / math.sqrt(20 + ((avg_Lp - 50) ** 2)))
    Sc = 1 + 0.045 * avg_Cp
    Sh = 1 + 0.015 * avg_Cp * T
    Rt = -math.sin(math.radians(2 * d_ro)) * Rc

    return math.sqrt(
        (dLp / Sl) ** 2 +
        (dCp / Sc) ** 2 +
        (dHp / Sh) ** 2 +
        Rt * (dCp / Sc) * (dHp / Sh)
    )

def build_perceptual_candidates(targets_df: pd.DataFrame, candidates_df: pd.DataFrame) -> pd.DataFrame:
    rows = []
    cand_records = candidates_df.to_dict("records")
    for t in targets_df.to_dict("records"):
        for c in cand_records:
            rows.append({
                "target_id": t["target_id"],
                "reference_id": c["id"],
                "de00_d50": ciede2000(t["lab_L"], t["lab_a"], t["lab_b"], c["lab_L"], c["lab_a"], c["lab_b"]),
                "lab_L": c["lab_L"],
                "lab_a": c["lab_a"],
                "lab_b": c["lab_b"],
                "hex": c.get("hex"),
                "is_atlas_only": c["is_atlas_only"],
                "source": c["source"],
                "passed_decision_gate": c["passed_decision_gate"],
            })
    return pd.DataFrame(rows)

decision_pool_df = atlas_df.copy()
perceptual_candidates_df = build_perceptual_candidates(targets_df, decision_pool_df)

if DECISION_MODE and FAIL_ON_NON_ATLAS_IN_DECISION:
    assert perceptual_candidates_df["is_atlas_only"].all()
    assert perceptual_candidates_df["passed_decision_gate"].all()
    assert set(perceptual_candidates_df["reference_id"]).issubset(atlas_reference_set)

per_target_ranked_df = (
    perceptual_candidates_df
    .sort_values(["target_id", "de00_d50", "reference_id"], ascending=[True, True, True])
    .groupby("target_id", group_keys=False)
    .head(TOP_K_PER_TARGET)
    .copy()
)

per_target_ranked_df["perceptual_rank"] = (
    per_target_ranked_df.groupby("target_id").cumcount() + 1
)

display(per_target_ranked_df.head(20))
```

## Structural metrics

This layer adds the native ARBE structure fields from the atlas and computes a deterministic structural decision order.

```python
STRUCTURAL_COLUMNS = [
    "id",
    "lambda_v2_nm",
    "lambda_ee_nm",
    "delta_lambda_nm",
    "mu2_nm2",
    "sigma_star_nm",
    "mu3_nm3",
]

per_target_ranked_df = per_target_ranked_df.merge(
    atlas_df[STRUCTURAL_COLUMNS + SCENARIO_DRIFT_COLUMNS + REFRESH_COLUMNS],
    left_on="reference_id",
    right_on="id",
    how="left",
    validate="many_to_one"
).drop(columns=["id"])

# Structural deltas are request-relative only if you have structural target values.
# In this production-ready baseline, we use the atlas-native delta_lambda_nm directly
# and keep mu2/sigma_star/mu3 as explicit structural descriptors.
per_target_ranked_df["delta_mu2_abs"] = per_target_ranked_df["mu2_nm2"].abs()
per_target_ranked_df["delta_sigma_star_abs"] = per_target_ranked_df["sigma_star_nm"].abs()
per_target_ranked_df["delta_mu3_abs"] = per_target_ranked_df["mu3_nm3"].abs()

# Scenario drift summary from the real atlas fields.
per_target_ranked_df["scenario_mean_drift"] = per_target_ranked_df[SCENARIO_DRIFT_COLUMNS].mean(axis=1)
per_target_ranked_df["candidate_max_drift"] = per_target_ranked_df[SCENARIO_DRIFT_COLUMNS].max(axis=1)
per_target_ranked_df["scenario_drift_spread"] = per_target_ranked_df[SCENARIO_DRIFT_COLUMNS].std(axis=1, ddof=0)

# Rank candidates structurally within each target.
per_target_ranked_df = per_target_ranked_df.sort_values(
    [
        "target_id",
        "delta_lambda_nm",
        "delta_mu2_abs",
        "delta_sigma_star_abs",
        "delta_mu3_abs",
        "de00_d50",
        "reference_id",
    ],
    ascending=[True, True, True, True, True, True, True]
).copy()

per_target_ranked_df["structural_rank"] = (
    per_target_ranked_df.groupby("target_id").cumcount() + 1
)

display(per_target_ranked_df.head(20))
```

## Group construction

Groups are generated only from atlas-approved candidates.  
The final decision remains **atlas-only**.

```python
candidate_lists = {
    t: per_target_ranked_df.loc[per_target_ranked_df["target_id"] == t, "reference_id"].tolist()
    for t in target_ids
}

group_rows = []
for refs in itertools.product(*candidate_lists.values()):
    if len(set(refs)) < len(refs):
        continue
    group_rows.append({
        "group_reference_ids": list(refs),
        "group_target_reference_pairs": list(zip(target_ids, refs)),
    })

groups_df = pd.DataFrame(group_rows)
print(f"group_count = {len(groups_df):,}")
display(groups_df.head())
```

```python
def materialize_group_rows(
    group_target_reference_pairs: list[tuple[str, str]],
    ranked_df: pd.DataFrame,
) -> pd.DataFrame:
    pair_df = pd.DataFrame(group_target_reference_pairs, columns=["target_id", "reference_id"])
    sub = pair_df.merge(
        ranked_df,
        on=["target_id", "reference_id"],
        how="left",
        validate="one_to_one",
    )

    if sub.isna().any(axis=None):
        missing_pairs = sub.loc[sub.isna().any(axis=1), ["target_id", "reference_id"]].drop_duplicates().to_dict("records")
        raise ValueError(f"Missing ranked rows for group pairs: {missing_pairs}")

    return sub

def score_group(group_target_reference_pairs: list[tuple[str, str]], ranked_df: pd.DataFrame) -> dict:
    sub = materialize_group_rows(group_target_reference_pairs, ranked_df)

    scenario_matrix = sub[SCENARIO_DRIFT_COLUMNS].to_numpy(dtype=float) if SCENARIO_DRIFT_COLUMNS else np.empty((len(sub), 0))
    scenario_per_candidate_mean = scenario_matrix.mean(axis=1) if scenario_matrix.size else np.full(len(sub), np.nan)
    scenario_per_candidate_max = scenario_matrix.max(axis=1) if scenario_matrix.size else np.full(len(sub), np.nan)
    scenario_per_candidate_spread = scenario_matrix.std(axis=1, ddof=0) if scenario_matrix.size else np.full(len(sub), np.nan)

    # Simple group geometry proxy in LAB space.
    labs = sub[["lab_L", "lab_a", "lab_b"]].to_numpy(dtype=float)
    centroid = labs.mean(axis=0)
    geometry_error = float(np.sqrt(((labs - centroid) ** 2).sum(axis=1)).mean())

    # Convert categorical refresh tiers to a stable lexical severity order.
    tier_cols = [c for c in sub.columns if c in ["refresh__firewall_calibrated_tier", "refresh__overall_calibrated_1_0b"]]
    tier_order = {"pass": 0, "review": 1, "warn": 2, "fail": 3}
    worst_tier = None
    if tier_cols:
        all_tiers = [str(x).strip().lower() for x in sub[tier_cols].fillna("").to_numpy().ravel() if str(x).strip()]
        if all_tiers:
            worst_tier = sorted(all_tiers, key=lambda x: tier_order.get(x, 99))[-1]

    return {
        "target_mean_de00": float(sub["de00_d50"].mean()),
        "target_max_de00": float(sub["de00_d50"].max()),
        "geometry_error": geometry_error,
        "scenario_mean_drift": float(np.nanmean(scenario_per_candidate_mean)),
        "mean_candidate_max_drift": float(np.nanmean(scenario_per_candidate_max)),
        "scenario_drift_spread": float(np.nanmean(scenario_per_candidate_spread)),
        "group_max_delta_lambda_nm": float(sub["delta_lambda_nm"].max()),
        "group_mean_delta_lambda_nm": float(sub["delta_lambda_nm"].mean()),
        "group_mean_mu2_abs": float(sub["delta_mu2_abs"].mean()),
        "group_mean_sigma_star_abs": float(sub["delta_sigma_star_abs"].mean()),
        "group_mean_mu3_abs": float(sub["delta_mu3_abs"].mean()),
        "delta_delta_lambda_nm": float(sub["delta_lambda_nm"].max() - sub["delta_lambda_nm"].min()),
        "worst_refresh_tier": worst_tier,
    }

scored_groups = []
for group_record in groups_df[["group_reference_ids", "group_target_reference_pairs"]].to_dict("records"):
    refs = group_record["group_reference_ids"]
    group_target_reference_pairs = group_record["group_target_reference_pairs"]
    scored = score_group(group_target_reference_pairs, per_target_ranked_df)
    scored_groups.append({
        "group_reference_ids": refs,
        "group_target_reference_pairs": group_target_reference_pairs,
        **scored,
        "all_atlas_only": all(r in atlas_reference_set for r in refs),
        "all_passed_decision_gate": True,
        "source": "enriched_json",
    })

group_scores_df = pd.DataFrame(scored_groups)
assert group_scores_df["all_atlas_only"].all()
assert group_scores_df["all_passed_decision_gate"].all()

display(group_scores_df.head())
```

## Deterministic decision order

Primary structural order:

1. `group_max_delta_lambda_nm`
2. `group_mean_mu2_abs`
3. `group_mean_sigma_star_abs`
4. `delta_delta_lambda_nm`

Perceptual terms remain visible and are used only after the structural priority keys.

```python
group_scores_df = group_scores_df.sort_values(
    [
        "group_max_delta_lambda_nm",
        "group_mean_mu2_abs",
        "group_mean_sigma_star_abs",
        "delta_delta_lambda_nm",
        "target_mean_de00",
        "target_max_de00",
        "geometry_error",
    ],
    ascending=[True, True, True, True, True, True, True]
).copy()

group_scores_df["decision_rank"] = np.arange(1, len(group_scores_df) + 1)

top_groups_df = group_scores_df.head(TOP_GROUPS_TO_EXPORT).copy()
display(top_groups_df.head(20))
```

## ARBE Structural Decision Criteria — first view

This section reports the **ARBE structural decision layer before perceptual interpretation**.

The goal is to make explicit **why** a group ranks first under ARBE Lambda Star logic:
- smallest feasible **group_max_delta_lambda_nm**
- lowest feasible **group_mean_mu2_abs**
- lowest feasible **group_mean_sigma_star_abs**
- lowest feasible **delta_delta_lambda_nm** as group-consistency review
- resulting **refresh robustness tier**

Only after this structural view should the perceptual D50 and scenario-drift metrics be interpreted.


```python
# Structural-first result view for the top-ranked groups
structural_first_cols = [
    "decision_rank",
    "group_reference_ids",
    "group_max_delta_lambda_nm",
    "group_mean_delta_lambda_nm",
    "group_mean_mu2_abs",
    "group_mean_sigma_star_abs",
    "group_mean_mu3_abs",
    "delta_delta_lambda_nm",
    "worst_refresh_tier",
    "target_mean_de00",
    "target_max_de00",
    "scenario_mean_drift",
    "mean_candidate_max_drift",
]

structural_first_groups_df = top_groups_df[structural_first_cols].copy()
display(structural_first_groups_df.head(20))
```

## Scenario-by-scenario illumination review

This section makes the **individual illumination / scenario results** visible for the top-ranked atlas-only group.

The underlying atlas stores separate scenario drift fields such as `de00_from_d50__ALS_BASE_D65`,
`de00_from_d50__ALS_BASE_A`, `de00_from_d50__ALS_LED_P1`, `de00_from_d50__ALS_LED_P2`,
`de00_from_d50__ALS_LED_P3`, `de00_from_d50__ALS_STR_1`, `de00_from_d50__ALS_STR_2`,
and `de00_from_d50__ALS_STR_3`.

The tables below show:
- the per-reference scenario drift values in wide form
- the same values in long form for technical reporting
- the worst scenario per selected reference


```python
if top_groups_df.empty:
    raise ValueError("No ranked groups available for scenario review.")

best_group_refs = top_groups_df.iloc[0]["group_reference_ids"]
best_group_target_reference_pairs = top_groups_df.iloc[0]["group_target_reference_pairs"]
top_group_detail_df = materialize_group_rows(best_group_target_reference_pairs, per_target_ranked_df)

scenario_display_cols = ["target_id", "reference_id", "de00_d50"] + SCENARIO_DRIFT_COLUMNS
available_refresh_cols = [c for c in ["refresh__overall_calibrated_1_0b", "refresh__firewall_calibrated_tier"] if c in top_group_detail_df.columns]
top_group_scenario_wide_df = top_group_detail_df[scenario_display_cols + available_refresh_cols].copy()

scenario_label_map = {c: c.replace("de00_from_d50__", "") for c in SCENARIO_DRIFT_COLUMNS}
top_group_scenario_wide_df = top_group_scenario_wide_df.rename(columns=scenario_label_map)

long_rows = []
for _, row in top_group_detail_df.iterrows():
    per_ref_pairs = [(scenario_label_map[c], float(row[c])) for c in SCENARIO_DRIFT_COLUMNS]
    worst_name, worst_value = max(per_ref_pairs, key=lambda x: x[1])
    for scenario_name, drift_value in per_ref_pairs:
        long_rows.append({
            "target_id": row["target_id"],
            "reference_id": row["reference_id"],
            "de00_d50": float(row["de00_d50"]),
            "scenario": scenario_name,
            "de00_from_d50": drift_value,
            "is_worst_scenario_for_reference": scenario_name == worst_name,
            "worst_scenario_for_reference": worst_name,
            "worst_scenario_drift_for_reference": worst_value,
        })

top_group_scenario_long_df = pd.DataFrame(long_rows)

per_reference_worst_scenario_df = (
    top_group_scenario_long_df
    .groupby(["target_id", "reference_id"], as_index=False)
    .agg(
        worst_scenario=("worst_scenario_for_reference", "first"),
        worst_scenario_drift=("worst_scenario_drift_for_reference", "first"),
        mean_scenario_drift=("de00_from_d50", "mean"),
        max_scenario_drift=("de00_from_d50", "max"),
        min_scenario_drift=("de00_from_d50", "min"),
    )
    .sort_values(["target_id", "reference_id"])
    .reset_index(drop=True)
)

print("Top-ranked group references:", best_group_refs)
display(top_group_scenario_wide_df)
display(per_reference_worst_scenario_df)
display(top_group_scenario_long_df.head(len(best_group_refs) * len(SCENARIO_DRIFT_COLUMNS)))
```
